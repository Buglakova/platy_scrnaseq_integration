```python
import pandas as pd
```

```python
smfish_input_df = pd.read_csv("../data/smfish/expression_matrix_aggregated_spots_spinning_disk_raw.tsv", sep="\t", index_col=0)
smfish_input_df.head()
```

```python
annotation_df = pd.read_csv("../data/smfish/sample_list_spinning_disk.csv", index_col=0)
annotation_df.head()
```

```python
annotation_df = annotation_df[(annotation_df["spotiflow_gene1"] == "yes") | (annotation_df["spotiflow_gene2"] == "yes")]
```

```python
gene_list = list(set(annotation_df.gene1.unique()) or set(annotation_df.gene2.unique()))
print(gene_list)
```

```python
smfish_filtered = smfish_input_df[gene_list]
smfish_filtered.head()
```

```python
smfish_filtered.to_csv("../data/smfish/expression_matrix_aggregated_spots_spinning_disk.tsv", sep="\t")
```
